## Part 2: Combining Conditions (30 minutes)

Combine multiple conditions to answer more specific questions.

**Rules:**
- Use `&` (AND), `|` (OR), `~` (NOT) - NOT `and`, `or`, `not`
- Always wrap each condition in parentheses

### Task 2.1: Two AND conditions
Find motorcycles (`MOTOCICLETA`) involved in fatal accidents (`CON MUERTOS`).

**Question:** How many fatal motorcycle accidents are in the dataset?

````python
# YOUR CODE HERE
# Pattern: df[(condition1) & (condition2)]
df_filtro5 = df[(df['tipo_vehiculo'] == 'MOTOCICLETA') & (df['gravedad_accidente'] == 'CON MUERTOS')]
print(f"Total fatal motorcycle accidents: {df_filtro5.shape[0]}")   
````

### Task 2.2: Three AND conditions
Find **AUTOMOVIL** vehicles, in **BOGOTA D.C.**, with severity **CON HERIDOS**.

Show the brand (`marca_vehiculo`) and model year (`modelo_vehiculo`) columns of the results.

````python
# YOUR CODE HERE


# Hint: after filtering, select columns with result[['marca_vehiculo', 'modelo_vehiculo']]
df_filtro6 = df[(df['tipo_vehiculo'] == 'AUTOMOVIL') & (df['gravedad_accidente'] == 'CON HERIDOS') & (df['departamento_accidente'] == 'BOGOTADC')]
df_filtro6[['marca_vehiculo', 'modelo_vehiculo']]
````

### Task 2.3: OR and .isin()
Find all accidents involving either **CHEVROLET**, **TOYOTA**, or **MAZDA** vehicles.

Try both approaches:
1. Using `|` (OR) operators
2. Using `.isin()`

Verify both give the same number of rows.

````python
# Approach 1: Using | (OR)
filtro_or = df[
    (df['marca_vehiculo'] == 'CHEVROLET') |
    (df['marca_vehiculo'] == 'TOYOTA') |
    (df['marca_vehiculo'] == 'MAZDA')
]

# Approach 2: Using .isin()
filtro_isin = df[
    df['marca_vehiculo'].isin(['CHEVROLET','TOYOTA','MAZDA'])
]

# Verify both have the same number of rows
print("OR:", len(filtro_or))
print("isin:", len(filtro_isin))
````

### Task 2.5: NOT with .isin()
Get all accident records EXCLUDING motorcycles (`MOTOCICLETA`) and buses (`BUS`).

**Pattern:** `df[~df['col'].isin([list])]`

Print the count and the remaining vehicle types.

````python
# YOUR CODE HERE
# After filtering, try: result['tipo_vehiculo'].value_counts() to see remaining types

resultado = df[
    ~df['tipo_vehiculo'].isin(['MOTOCICLETA','BUS'])
]

print("Cantidad:", len(resultado))

print("\nTipos restantes:")
print(resultado['tipo_vehiculo'].value_counts())
````

## Part 3: Convenience Methods (20 minutes)

Practice .between(), .str.contains(), .str.startswith(), and .query() for common filtering patterns.

### Task 3.1: .between() for model year
Find vehicles with model year between **2015 and 2020** (inclusive).

**Pattern:** `df[df['col'].between(low, high)]`

Print how many records match and show the distribution of model years in the result.

````python
# YOUR CODE HERE
# Hint: after filtering, use result['modelo_vehiculo'].value_counts().sort_index()
resultado = df[df['modelo_vehiculo'].between(2015, 2020)]

print("Cantidad:", len(resultado))

print(resultado['modelo_vehiculo'].value_counts().sort_index())
````

### Task 3.2: .between() for age range
Get vehicles aged **0 to 5 years** (relatively new vehicles).

Print the count and the percentage of the total dataset.

````python
# YOUR CODE HERE
resultado = df[df['edad_vehiculo'].between(0,5)]

print("Cantidad:", len(resultado))

print("Porcentaje:", len(resultado)/len(df)*100, "%")
````

### Task 3.3: .str.contains() for text search
Find transit authorities (`autoridad_de_transito`) whose name contains **"BOGOTA"**.

Print the unique authority names that match.

**Remember:** Always add `na=False` to handle potential NaN values.

````python
# YOUR CODE HERE
# Pattern: df[df['col'].str.contains('text', na=False)]
resultado = df[
    df['autoridad_de_transito'].str.contains('BOGOTA', na=False)
]

print(resultado['autoridad_de_transito'].unique())
````

### Task 3.4: .str.startswith()
Find vehicle brands that start with **"CH"** (Chevrolet, Changan, Chery, etc.).

Print the unique brand names that match and the total number of records.

**Pattern:** `df[df['col'].str.startswith('text', na=False)]`

````python
# YOUR CODE HERE
resultado = df[
    df['marca_vehiculo'].str.startswith('CH', na=False)
]

print("Marcas:")
print(resultado['marca_vehiculo'].unique())

print("\nCantidad:", len(resultado))
````

### Task 3.5: .query() method
Rewrite this boolean indexing filter using `.query()`:

```python
df[(df['edad_vehiculo'] > 10) & (df['departamento_accidente'] == 'ANTIOQUIA') & (df['tipo_vehiculo'] == 'MOTOCICLETA')]
```

Verify both approaches give the same number of rows.

````python
# Boolean indexing version (given):
result_bool = df[
    (df['edad_vehiculo'] > 10) &
    (df['departamento_accidente'] == 'ANTIOQUIA') &
    (df['tipo_vehiculo'] == 'MOTOCICLETA')
]

# YOUR CODE HERE - .query() version:
# result_query = df.query('...')

# Verify:
# print(f"Boolean indexing: {len(result_bool)} rows")
# print(f".query(): {len(result_query)} rows")

result_query = df.query(
    "edad_vehiculo > 10 and departamento_accidente == 'ANTIOQUIA' and tipo_vehiculo == 'MOTOCICLETA'"
)

print(f"Boolean indexing: {len(result_bool)} rows")
print(f".query(): {len(result_query)} rows")
````

### Reflection 1
Which filtering method did you find most useful: boolean indexing, .isin(), .between(), .str.contains(), or .query()? Why?

**Your answer:**

The most useful method depends on the context: for simple filters with operators `&`, `|`, `~`, it's clear and straightforward. However, `.isin()` is especially useful when you need to filter by multiple values ​​in a list (such as specific vehicle makes), because it's more readable than writing multiple OR conditions. `.between()` is ideal for numeric ranges (age, vehicle model) and is an elegant way to express bound inclusion. `.query()` is excellent for complex queries because the syntax is more similar to natural language, which improves code readability.

### Reflection 3
Why is it important to clean the data (Week 3) BEFORE filtering it (Week 4)? What problems could arise if you filter dirty data?

**Your answer:**


1. **Inconsistencies in values**: If department names vary (e.g., 'BOGOTA', 'BOGOTA D.C.', 'Bogota'), your filters may miss records. This would lead to incomplete analysis and biased conclusions about accidents in those regions.

2. **NULL/NaN values**: If you have missing values ​​in filter columns (such as `accident_department`), records may be inadvertently excluded even if they are relevant, resulting in data loss.

3. **Incorrect data types**: If `vehicle_age` is loaded as text instead of a number, the `vehicle_age > 15` comparison would fail or be incorrect, affecting analyses of older vehicles.

4. **Duplicates**: Duplicate records would inflate counts and statistics, making numbers like "758 motorcycles in fatal accidents" misleading.

5. **Special characters and spaces**: Values ​​like 'MOTORCYCLE' (with a space) would not match 'MOTORCYCLE' in your filter, fragmenting your results.
